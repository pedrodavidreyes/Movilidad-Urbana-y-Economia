
### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

```python

# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic["update_time_utc"] = pd.to_datetime(traffic["update_time_utc"], errors="coerce", utc=True)
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic["update_time_utc_week_ago"], errors="coerce", utc=True)
# verificar el cambio
traffic.info()

```

```python
# Limpia separadores y convierte columnas numéricas en eco
eco['city_gdp_capita'] = eco['city_gdp_capita'].astype(str).str.replace('.', '').str.replace(',', '.').astype(float)
# unemployment es diferente al caso anterior y el codigo se rompe
# no tiene comas, en su lugar tiene %
# el punto esta donde debe estar por lo tanto no hay que hacer nada al respecto con el punto
eco['unemployment_pct'] = eco['unemployment_pct'].astype(str).str.replace('%', '').astype(float)
# population_m tambien es diferente
# el punto esta donde debe estar por lo tanto no hay que hacer nada al respecto
eco['population_m'] = eco['population_m'].astype(str).str.replace(',', '.').astype(float)

# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco["population_m"]*1000000

# verificar el cambio
eco.info()
eco.head(3)
# por ultimo estamos suponiendo que no hay espacios en blanco y que no capturaron caracteres raros
# no lo pidieron pero y que pasa con las pm2.5??? sigue en formato erroneo
```

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python

# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_kms','jams_count','mins_delay',
             'travel_time_live_per_10kms_mins','travel_time_hist_per_10kms_mins']

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = pd.merge(traffic_2024_small, eco_2024_small, on=["city","year"], how="inner")
# Mostrar las primeras 5 filas

merged.head(5)
```

```python
# Crear boxplot para observar el comportamiento de los minutos de congestion JamsDelay
# crea tu gráfico
# Me tome la libertad de ordenarlo para facilitar la lectura del grafico

#creamos una lista ordenada de ciudades, de mayor a menor promedio
order = (
    traffic_2024
    .groupby("city")["jams_delay"]
    .mean()
    .sort_values(ascending=False)
    .index
)

top10=order[:10]  # Solo los diez mas altos, porque al final vamos a correlacionar si el trafico afecta el gdp

# Es neceario cambiar el tamaño para que sea legible
plt.figure(figsize=(12, 4))

sns.boxplot(
    data=traffic_2024,
    x="jams_delay",
    y="city",
    order=top10   # en caso que se quieran todos los datos, solo se quita esta linea
)

# se añade esta linea para que los nombres de las ciudades no queden amontonadas
plt.tight_layout()

# obtener promedio para mostrarlo en título
mean_value = merged['jams_delay'].mean()
plt.title(f'Boxplot de JamsDelay (2024)\nPromedio: {mean_value:.2f}')
plt.show()

```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)
eco_2024_small["city_gdp_capita"].hist(bins=10)
plt.title('Distribución de la economia PIB (2024)')

plt.xlabel('PIB per cápita')
plt.ylabel('Frecuencia')

plt.show()
```

```python
# Gráfico de barras para comparar jams_delay y city_gdp_capita por ciudad
#merged.plot( ... , y=['jams_delay', 'city_gdp_capita'])

# Al realizarlo como se sugiere hay un error, estamos comparando variables con unidades distintas, minutos contra ingreso por persona
#merged_jam_asc= merged.sort_values(by="jams_delay", ascending=False)
#merged_jam_asc.plot(
#    kind="bar",
#    x="city",
#    y=["jams_delay","city_gdp_capita"])
#plt.xticks(rotation=90)
#plt.title("Comparativo Trafico vs Economia")
#plt.tight_layout()
#plt.show()
# por lo tanto este grafico es engañoso, seria mejor usar porcentajes en base el valor mas alto.

#base 0 a 1
merged_norm = merged.copy()
merged_norm["jams_delay_norm"] = (merged_norm["jams_delay"] / merged_norm["jams_delay"].max())
merged_norm["city_gdp_capita_norm"] = (merged_norm["city_gdp_capita"] / merged_norm["city_gdp_capita"].max())
#ordenamos por trafico
#merged_norm = merged_norm.sort_values(by="jams_delay_norm", ascending=False)
# ordenamos por pib
merged_norm = merged_norm.sort_values(by="city_gdp_capita_norm", ascending=False)

#parametros del grafico
merged_norm.plot(
    kind="bar",
    x="city",
    y=["jams_delay_norm", "city_gdp_capita_norm"],
    figsize=(12, 5)
)

plt.xticks(rotation=90)
plt.title("Comparativo normalizado: Tráfico vs Economía (2024)")
plt.ylabel("Valor normalizado (0–1)")
plt.tight_layout()
plt.show()
```

# 🧾 Resumen ejecutivo

**Contexto & objetivo:**  
El análisis evalúa la relación entre la movilidad urbana y la productividad económica (PIB per cápita) en 15 ciudades de 7 países de América Latina durante 2024. Se utilizaron como variables clave: el retraso total por congestión (jams_delay) y el PIB per cápita, con el objetivo de identificar las ciudades prioritarias para invertir en infraestructura de transporte.

**Cobertura de datos:**  
Los datos de movilidad y economía provienen de fuentes distintas (Tom Tom Traffic Index y OECD Cities) y se integraron a nivel ciudad-año. Se verificó la consistencia temporal y geográfica entre datasets. El análisis integra más de 1 millón de registros de movilidad urbana con datos económicos de 7 países (Argentina, Brasil, Chile, Colombia, México, Perú, Uruguay).

**Metodología (alto nivel):**  
Se realizarón procesos de limpieza, estandarización (renombrado de columnas y correción de formatos númericos y de fecha) y agregación de datos. se creó columna de año el en dataset de movilidad, se filtraron ambos datasets para el año 2024 y se calcularon promedios a nivel ciudad. Por seguridad, se crearon subconjuntos para evitar modificar el dataset original. Se utilizó una unión tipo INNER para combinar ambos datasets (por ciudad y año), y se aplicaron visualizaciones exploratorias (boxplots, histogramas y gráficos comparativos) para analizar distribuciones, detectar outliers y evaluar tendencias.
Para facilitar la comparación entre métricas de distinta escala, se utilizó una normalización basada en valores máximos.

**Hallazgos iniciales:**  
Se observó que no existe una relación directa entre el nivel de congestión y el PIB per cápita. Ciudad de México domina claramente con el tráfico mas alto (jams_delay: 2833.06 el más alto de la muestra) junto con un alto PIB per cápita, mientras que Montevideo muestra el caso opuesto, con bajo tráfico (jams_delay: 50.20 el más bajo de la muestra) y el mayor PIB per cápita.
Además se observa alta dispersión en casi todas las ciudades, con outliers que superan los 10,000 minutos, lo que indica que el problema real está en los picos, no en el promedio.<br>
A causa de la dispersión los promedios pueden ser engañosos, se recomienda utilizar métricas robustas como la mediana, para análisis posteriores

**Recomendaciones**  
Se recomienda priorizar la inversión en infraestructura de transporte en Bogotá (traffic_index_live: 37.61 el más alto de la muestra) y, en segundo lugar en Lima, debido a que son ciudades con tráfico alto y bajo PIB per cápita. Las acciones sugeridas incluyen la expansión del transporte masivo (metro), la implementación de semaforización inteligente y la creación de carriles exclusivos para autobuses.<br>
Para ciudades con tráfico alto y PIB alto, como Ciudad de México y Sao Paulo, se sugiere reducir el tráfico total, mediante estrategias de gestión de la demanda (horarios escalonados, teletrabajo, peajes urbanos en horas pico y mejorar la logística urbana (carga/descarga).<br>
Asimismo replicar las buenas prácticas de ciudades eficientes como Montevideo, Buenos aires y Brasilia con un alto PIB y bajo tráfico, mediante el fortalecimiento de políticas de planeación urbana y distribución modal, la adopción de estándares de interoperabilidad de transporte, la gestión de flotas y el fomento de la micromobilidad.<br>
